```python
import pandas as pd
import random
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime
```

```python
feedback_df = pd.read_excel("lawish_customer_feedback_responses.xlsx")

sns.set(style="whitegrid")
```

```python
satisfaction_order = ['Very Unsatisfied', 'Unsatisfied', 'Neutral', 'Satisfied', 'Very Satisfied']

plt.figure(figsize=(6, 4))
ax = sns.countplot(data=feedback_df, x='Overall Satisfaction', order=satisfaction_order, hue='Overall Satisfaction', palette='viridis')
plt.title('Overall Customer Satisfaction')
plt.xticks(rotation=45)
plt.xlabel('Satisfaction Level')
plt.ylabel('Number of Responses')
# Add counts on top of each bar
for p in ax.patches:
    ax.annotate(
        format(p.get_height(), '.0f'),
        (p.get_x() + p.get_width() / 2., p.get_height()),
        ha = 'center', va = 'center',
        xytext = (0, 5),
        textcoords = 'offset points'
    )
plt.show()
```

```python
service_feedback_counts = feedback_df['Service Feedback'].value_counts()
plt.figure(figsize=(5, 5))
plt.pie(service_feedback_counts, labels=service_feedback_counts.index, autopct='%1.1f%%', startangle=140)
plt.title('Service Feedback Distribution')
plt.show()
```

```python
feedback_mapping = {
    'Early': 0,
    'On time': 1,
    '1-2 days late': 2,
    '3-5 days late': 3,
    'More than a week late': 4
}

feedback_df['Delivery Time Feedback Numerical'] = feedback_df['Delivery Time Feedback'].map(feedback_mapping)

plt.figure(figsize=(5, 5))
sns.boxplot(data=feedback_df, y='Delivery Time Feedback Numerical')
plt.title('Delivery Time Feedback')
plt.yticks(np.arange(5), list(feedback_mapping.keys()))
plt.xlabel('Delivery Time Feedback')
plt.ylabel('Delivery Time Feedback')
plt.show()
```

```python
staff_behavior_counts = feedback_df['Staff Behavior'].value_counts()
plt.figure(figsize=(5, 5))
plt.pie(staff_behavior_counts, labels=staff_behavior_counts.index, autopct='%1.1f%%', startangle=90)
plt.title('Staff Behavior Feedback Distribution')
plt.axis('equal')
plt.show()
```

```python
store_experience_order = ['Very clean and organized', 'Clean but crowded', 'Average', 'Difficult to find products', 'Messy and unorganized']
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=feedback_df, y='Store Experience', order=store_experience_order, hue='Store Experience', palette='viridis')
plt.title('Store Experience Feedback')
plt.xlabel('Number of Responses')
plt.ylabel('Store Experience')
# Add counts on top of each bar
for p in ax.patches:
    ax.annotate(
        format(p.get_width(), '.0f'),
        (p.get_width(), p.get_y() + p.get_height() / 2),
        ha='center', va='center',
        xytext=(5, 0),
        textcoords='offset points'
    )
plt.show()
```

```python
collection_order = ['Wide variety', 'Good variety', 'Average collection', 'Limited choices', 'Poor collection']
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=feedback_df, y='Variety/Collection Feedback', order=collection_order, hue='Variety/Collection Feedback', palette='plasma')
plt.title('Variety/Collection Feedback')
plt.xlabel('Number of Responses')
plt.ylabel('Variety/Collection Feedback')
# Add counts on top of each bar
for p in ax.patches:
    ax.annotate(
        format(p.get_width(), '.0f'),
        (p.get_width(), p.get_y() + p.get_height() / 2),
        ha='center', va='center',
        xytext=(5, 0),
        textcoords='offset points'
    )
plt.show()
```

```python
price_satisfaction_order = [
    'Very Affordable',
    'Reasonable',
    'Average',
    'Expensive',
    'Very Expensive'
]
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=feedback_df, y='Price Satisfaction', hue='Price Satisfaction', palette='viridis', order=price_satisfaction_order)
plt.title("Distribution of Price Satisfaction Ratings", fontsize=16)
plt.xlabel("Number of Feedbacks", fontsize=14)
plt.ylabel("Price Satisfaction Rating", fontsize=14)
# Add count annotations on each bar
for p in ax.patches:
    ax.annotate(
        format(p.get_width(), '.0f'),
        (p.get_width(), p.get_y() + p.get_height() / 2),
        ha='center', va='center',
        xytext=(5, 0),
        textcoords='offset points'
    )
plt.show()
```

```python
feedback_scores = {
    'Service Feedback': {
        '5 - Excellent': 5,
        '4 - Very Good': 4,
        '3 - Satisfactory': 3,
        '2 - Needs Improvement': 2,
        '1 - Poor': 1
    },
    'Delivery Time Feedback': {
        'Early': 5,
        'On time': 4,
        '1-2 days late': 3,
        '3-5 days late': 2,
        'More than a week late': 1
    },
    'Staff Behavior': {
        'Very polite': 5,
        'Helpful': 4,
        'Neutral': 3,
        'Not helpful': 2,
        'Rude': 1
    },
    'Store Experience': {
        '5 - Very clean and organized': 5,
        '4 - Clean but crowded': 4,
        '3 - Average': 3,
        '2 - Difficult to find products': 2,
        '1 - Messy and unorganized': 1
    },
    'Variety/Collection Feedback': {
        '5 - Wide variety': 5,
        '4 - Good variety': 4,
        '3 - Average collection': 3,
        '2 - Limited choices': 2,
        '1 - Poor collection': 1
    },
    'Price Satisfaction': {
        '5 - Very Affordable': 5,
        '4 - Reasonable':4,
        '3 - Average': 3,
        '2 - Expensive': 2,
        '1 - Very Expensive':1
    },
    'Overall Satisfaction': {
        '5 - Very Satisfied': 5,
        '4 - Satisfied': 4,
        '3 - Neutral': 3,
        '2 - Unsatisfied': 2,
        '1 - Very Unsatisfied': 1
    }
}

for column, score_map in feedback_scores.items():
    feedback_df[f"{column} Score"] = feedback_df[column].map(score_map)
```

```python
# MEAN
mean_service_feedback = feedback_df['Service Feedback Score'].mean()
# mean_delivery_time_feedback = feedback_df['Delivery Time Feedback Score'].mean()
mean_staff_behavior = feedback_df['Staff Behavior Score'].mean()
mean_store_experience = feedback_df['Store Experience Score'].mean()
mean_variety_feedback = feedback_df['Variety/Collection Feedback Score'].mean()
mean_price_satisfaction = feedback_df['Price Satisfaction Score'].mean()
mean_overall_satisfaction = feedback_df['Overall Satisfaction Score'].mean()
print("Mean Service Feedback:", mean_service_feedback)
# print("Mean Delivery Time Feedback:", mean_delivery_time_feedback)
print("Mean Staff Behavior:", mean_staff_behavior)
print("Mean Store Experience:", mean_store_experience)
print("Mean Variety Feedback:", mean_variety_feedback)
print("Mean Price Satisfaction Feedback:", mean_price_satisfaction)
print("Mean Overall Satisfaction:", mean_overall_satisfaction)
```
